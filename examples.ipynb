```python

import math
from qiskit import QuantumCircuit
from qiskit2quantikz import CircuitDrawing, rewrite_rule, qiskit_rewrite
```

```python
# If you want to see the results in this notebook
import pymupdf
from IPython.display import Image
def display_as_png(filename : str, zoom=2):
    base = filename.split(".")[0]
    pdf_document = pymupdf.open(filename)
    page = pdf_document.load_page(0)
    pix=page.get_pixmap(matrix=pymupdf.Matrix(zoom, zoom))
    pix.save(f"{base}.png")
    display(Image(filename=f"{base}.png"))
```

```python
qft_circuit = QuantumCircuit(4)

qft_circuit.h(0)
qft_circuit.cp(math.pi/2, 0, 1)
qft_circuit.cp(math.pi/4, 0, 2)
qft_circuit.cp(math.pi/8, 0, 3)

qft_circuit.h(1)

qft_circuit.cp(math.pi/2, 1, 2)
qft_circuit.cp(math.pi/4, 1, 3)
qft_circuit.h(2)
qft_circuit.cp(math.pi/2, 2, 3)
qft_circuit.h(3)
d1 = CircuitDrawing(qft_circuit)
d1.draw(standalone=True,filename="circ3.tex", qubit_labels="q_n")
```

```python
c1 = QuantumCircuit(5)
c2 = QuantumCircuit(5)
for i in range(1,5):
    c1.cx(i,0)
c1.cx(1,0)
for i in range(2, 5):
    c2.cx(i,0)
d1 = CircuitDrawing(c1)
d2 = CircuitDrawing(c2)
s = rewrite_rule(d1, d2, standalone=True, filename="rule1.tex")
```

```python
d1 = CircuitDrawing(4)


d1.cx(0,1)
d1.cx(2,3)
d1.cx(1,2)
d1.draw(standalone=True,filename="circ3.tex", qubit_labels=["q0","q1","q2", "q3"])
```

```python
!pdflatex circ3.tex
display_as_png("circ3.pdf")
```

```python
c1 = QuantumCircuit(5)
for i in range(1,5):
    c1.cx(i,0)
c1.cx(1,0)
s = qiskit_rewrite(c1, standalone=True, filename="rule1_alt.tex")
```

```python
!latex rule1.tex
display_as_png("rule1_alt.pdf")
```

```python
c1 = QuantumCircuit(2)
s1 = QuantumCircuit(2)
c2 = QuantumCircuit(2)
s2 = QuantumCircuit(1)

c1.h(1)
s1.rz(math.pi/2, 0)
s1.cx(0,1)
d1 = CircuitDrawing(c1)
d1.add_boxed_subcirc(s1,[0,1])
d1.rz(math.pi/2,0)

c2.h(1)
c2.cx(0,1)
s2.rz(math.pi/2,0)
s2.rz(math.pi/2,0)
d2 = CircuitDrawing(c2)
d2.add_boxed_subcirc(s2, [0])

s = rewrite_rule(d1, d2, standalone=True, filename="rule2.tex")
```

```python
!pdflatex rule2.texA
display_as_png("rule2.pdf")
```

```python

# alternatively...
d1 = CircuitDrawing(2)
d1.h(1)
d1.begin_box([0,1])
d1.rz(math.pi/2, 0)
d1.cx(0,1)
d1.end_box()
d1.rz(math.pi/2,0)

d2 = CircuitDrawing(2)
d2.h(1)
d2.cx(0,1)
d2.begin_box([0])
d2.rz(math.pi/2,0)
d2.rz(math.pi/2,0)
d2.end_box()

s = rewrite_rule(d1, d2, standalone=True, filename="rule2_alt.tex")
```

```python
!pdflatex rule2_alt.tex
display_as_png("rule2_alt.pdf")
```

```python
d1 = CircuitDrawing(2)
d1.swap(0,1)
d2 = CircuitDrawing(2)
d2.cx(0,1)
d2.cx(1,0)
d2.cx(0,1)
rewrite_rule(d1, d2, standalone=True, filename="swap_rule.tex")
```

```python
!pdflatex swap_rule.tex
display_as_png("swap_rule.pdf")
```

```python
d = CircuitDrawing(1)
d.rz("\\theta/2", 0)
d.draw(standalone="True", filename="symb_angle.tex")
!pdflatex symb_angle.tex
display_as_png("symb_angle.pdf")
```

```python
d = CircuitDrawing(2)
d.h(1)
d.cx(0,1)
d.begin_box([0,1])
d.t(0)
d._circuit.y(0)
d._circuit.y(1)
d.s(1)
d.end_box()
d.tdg(1)
d._circuit.y(1)
d.draw(standalone=True, filename="t-gates.tex", qubit_labels="q_n")
```

```python
!pdflatex t-gates.tex
display_as_png("t-gates.pdf")
```

```python
d = CircuitDrawing(4)
d.h(q for q in range(4))
d.block("U_f", 0, 3)
d.h(q for q in range(4))
d.draw(standalone=True, filename="block.tex", qubit_labels=["\ket{0}" for _ in range(4)])
!pdflatex block.tex
display_as_png("block.pdf")
```

```python
d = CircuitDrawing(4)
d.block("R_{ZXZX}(\\theta)", 0, 3)
d.draw(standalone=True, filename="paulistr.tex")
d1 = CircuitDrawing(4)
d1.h(1)
d1.h(2)
d1.cx(0,1)
d1.cx(1,2)
d1.cx(2,3)
d1.rz("\\theta", 3)
d1.cx(2,3)
d1.cx(1,2)
d1.cx(0,1)
d1.h(1)
d1.h(2)


rewrite_rule(d, d1, standalone=True, filename="paulistr_rule.tex")
```
